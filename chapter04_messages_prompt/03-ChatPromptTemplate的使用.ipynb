### ChatPromptTemplate 的使用

#### 1. ChatPromptTemplate 的实例化

#### 1.1 from_messages()

```python
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "你是一个友好的 AI 助手, 你的名字叫{name}"),
    ("human", "你好, 最近怎么样?"),
    ("ai", "我很好, 谢谢"),
    ("human", "{user_input}")
])

result = chat_prompt_template.invoke({"name": "小智", "user_input": "2+2=?"})
rprint(result)
```

传入参数: 字典列表: 返回值类型 ChatPromptValue

```python
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "你是一个友好的 AI 助手, 你的名字叫{name}"),
    ("human", "你好, 最近怎么样?"),
    ("ai", "我很好, 谢谢"),
    ("human", "{user_input}")
])

result = chat_prompt_template.invoke({"name": "小智", "user_input": "2+2=?"})
rprint(result)
rprint(type(result))
```

#### 2.2 format() 参数类型: 变量值: 返回值类型: 字符串

### 结合大模型的调用

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from rich import print as rprint
from langchain_core.prompts import ChatPromptTemplate

load_dotenv(override=True)
ZHIPUAI_API_KEY = os.getenv("ZHIPUAI_API_KEY")
ZHIPUAI_API_BASE_URL = os.getenv("ZHIPUAI_API_BASE_URL")

# 获取大模型
model = init_chat_model(
    model='glm-5.3-flash',
    model_provider="openai",
    api_key=ZHIPUAI_API_KEY,
    base_url=ZHIPUAI_API_BASE_URL
)


chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "你是一个友好的 AI 助手, 你的名字叫{name}"),
    ("human", "你好, 最近怎么样?"),
    ("ai", "我很好, 谢谢"),
    ("human", "{user_input}")
])

prompt_value = chat_prompt_template.format_messages(name ="小智", user_input= "2+2=?")
response = model.invoke(prompt_value)

rprint(response)
```

### 2.4 更丰富的初始化参数类型

#### 4.1 字符串列表

```python
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint
chat_prompt_template = ChatPromptTemplate.from_messages([
    "你好, 我是 {name}"  # 这里会自动变成 HumanMessage
])
response = chat_prompt_template.invoke({"name": "小明"})
rprint(response)
```

#### 4.2 元组列表

```python
chat_prompt_template = ChatPromptTemplate.from_messages([
    ('human', '你好, 我是{name}')
])

response = chat_prompt_template.invoke({"name": "小明"})
rprint(response)
```

#### 4.3 字典列表

```python
chat_prompt_template = ChatPromptTemplate.from_messages([
    {
        "role": "system",
        "content": "你是一个友好的 AI 助手"
    },
    {
        "role": "human",
        "content": "你好, 我是{name}"
    }
])

response = chat_prompt_template.invoke({"name": "小明"})
rprint(response)
```

#### 4.4 消息对象列表

```python
from langchain_core.messages import SystemMessage, HumanMessage

chat_prompt_template = ChatPromptTemplate.from_messages([
    SystemMessage("你是一个友好的 AI 助手"),
    HumanMessage("你好, 我是{name}")
])

response = chat_prompt_template.invoke({"name": "小明"})
rprint(response)
```

#### BaseMessagePromptTemplate 参数列表

```python
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_core.prompts import SystemMessagePromptTemplate, HumanMessagePromptTemplate

system_mess_template = SystemMessagePromptTemplate.from_template("你是一个友好的 AI 助手")
human_mess_template = HumanMessagePromptTemplate.from_template("你好, 我是{name}")

chat_prompt_template = ChatPromptTemplate.from_messages([
    system_mess_template,
    human_mess_template
])

response = chat_prompt_template.invoke({"name": "小明"})
rprint(response)
```
