### 输入的格式化

举例1: 保存多模态的数据, 使用字典列表

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from rich import print as rprint
import base64

load_dotenv(override=True)
ZHIPUAI_API_KEY = os.getenv("ZHIPUAI_API_KEY")
ZHIPUAI_API_BASE_URL = os.getenv("ZHIPUAI_API_BASE_URL")

# 获取大模型
model = init_chat_model(
    model='glm-5.3-flash',
    model_provider="openai",
    api_key=ZHIPUAI_API_KEY,
    base_url=ZHIPUAI_API_BASE_URL
)

def encode_image(img_path, img_type='jpeg'):
    '''将一张本地图片转换成 Base64 编码的 Data URI 字符串, 方便在文本中嵌入图片数据'''
    with open(img_path, 'rb') as img_file:
        # return f"data:image/{img_type};base64,{base64.b64encode(img_file.read()).decode('utf-8')}"
        return base64.b64encode(img_file.read()).decode('utf-8')

img_path = "../images/RAG.png"

base64_image = encode_image(img_path)

print(base64_image)

response = model.invoke(
    [
        # 此种格式可用
        # HumanMessage(
        #     content = [
        #         {
        #             "type": "text",
        #             "text": "这张图里有什么"
        #         },
        #         {
        #             "type": "image_url",
        #             "image_url":  {
        #                 "url": base64_image
        #             }
        #         }
        #     ]
        # )
        # 推荐统一写法
        HumanMessage(
            content_blocks = [
                {
                    "type": "text",
                    "text": "这张图里有什么"
                },
                {
                    "type": "image",
                    "base64": base64_image,
                    "mime_type": "image/png"
                }
            ]
        )
    ]
)
rprint(response.content)
```

### 统一输出格式

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from rich import print as rprint
import base64

load_dotenv(override=True)
ZHIPUAI_API_KEY = os.getenv("ZHIPUAI_API_KEY")
ZHIPUAI_API_BASE_URL = os.getenv("ZHIPUAI_API_BASE_URL")

# 获取大模型
model = init_chat_model(
    model='glm-5.3-flash',
    model_provider="openai",
    api_key=ZHIPUAI_API_KEY,
    base_url=ZHIPUAI_API_BASE_URL
)

response = model.invoke('你好, 一句话回答')
print('='*20, '-> response <-', '='*20)
print(response)
print('='*20, '-> response.content <-', '='*20)
print(response.content)
print('='*20, '-> response.content_blocks <-', '='*20)
print(response.content_blocks)
```
