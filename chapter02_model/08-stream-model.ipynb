#### 流式调用(stream)

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

ZHIPUAI_API_KEY = os.getenv("ZHIPUAI_API_KEY")
ZHIPUAI_API_BASE_URL = os.getenv("ZHIPUAI_API_BASE_URL")

# 获取大模型
model = init_chat_model(
    model='glm-5.3',
    model_provider="openai",
    api_key=ZHIPUAI_API_KEY,
    base_url=ZHIPUAI_API_BASE_URL,
    max_tokens=10
)
```

```python
for chunk in model.stream('帮我解释一下什么是人工智能?'):
    print(chunk.text, end='', flush=True)
```

#### 批量调用(batch)

##### batch()一次性接收所有响应

```python
message = [
    "你是谁",
    "2+3*5=?",
    "中国首都在哪里?"
]
responses = model.batch(message)
for response in responses:
    print(response)
```

##### batch_as_completed() 一次性接收所有响应

```python
message = [
    "你是谁",
    "2+3*5=?",
    "中国首都在哪里?"
]
responses = model.batch_as_completed(message)
for response in responses:
    print(response)
```

```python
# 准备多个输入
inputs = [
    '翻译成英文: 春天来了',
    '翻译成英文: 夏天很热',
    '翻译成英文: 秋天落叶',
    '翻译成英文: 冬天下雪'
]
# 批量调用(高效)
import time
start = time.time()
responses = model.batch(inputs)
batch_time = time.time() - start

print('批量调用结果: ')
for i, response in enumerate(responses):
    print(f'{i+1}. {response.content}')
print(f'耗时: {batch_time: .2f}秒\n')
```
