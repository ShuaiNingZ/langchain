1.3 通过阿里云百炼平台获取相关的大模型

```python
import os

from dotenv import load_dotenv
from langchain_community.chat_models import ChatTongyi

# 1. 读取 .env 配置文件中的信息, 相关的环境变量以 .env 文件中的优先
load_dotenv(override=True)

api_key = str(os.getenv('DASHSCOPE_API_KEY'))
DASHSCOPE_API_BASE_URL = os.getenv('DASHSCOPE_API_BASE_URL')

# 2. 模型的初始化
llm = ChatTongyi(
    model='qwen-plus',
    api_key=api_key,
    api_base=DASHSCOPE_API_BASE_URL
)

# 3. 模型的调用
response = llm.invoke('请用一句话介绍你自己')
print(response)
```
